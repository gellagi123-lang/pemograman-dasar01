```python
food = "Mi ayam bakso"
jumlah_porsi = 2
harga_satuan = 15000.500
kode_transaksi = 255
nomor_meja = 8

# 1. Operator %s untuk string (teks)
print("Hendi sedang memesan %s" % food)

# 2. Operator %d untuk integer (angka bulat)
print("Jumlah porsi yang dipesan ada %d mangkok" % jumlah_porsi)

# 3. Operator %f untuk float (angka desimal)
print("Harga satu porsi adalah %.3f rupiah" % harga_satuan)

# 4. Operator %x untuk heksadesimal
print("Kode unik transaksi (%x)" % kode_transaksi)

# 5. Operator %o untuk oktal
print("Nomor meja pesanan (%o)" % nomor_meja)
```

```python
# 1. Kalkulator sederhana
a = 20
b = 5
print("Penjumlahan:", a + b)
print("Pengurangan:", a - b)
print("Perkalian:", a * b)
print("Pembagian:", a / b)

# 2. Kalkulator kasir
harga_makan = int(input("Berapa harga mi ayam per porsi? "))
jumlah_makan = int(input("Berapa porsi yang Anda beli? "))
total_makan = harga_makan * jumlah_makan
print(f"Harga makan per porsi = {harga_makan}, totalnya adalah {total_makan}")

# 3. Mengecek bilangan
bilangan = 8
hasil = (bilangan > 0) and (bilangan % 2 == 0)
print("Genap dan positif?", hasil)
```

```python
# konversi waktu
total_detik = int(input("Masukkan total detik: "))
jam = total_detik // 3600
sisa_detik = total_detik % 3600
menit = sisa_detik // 6
detik = sisa_detik % 60
print(f"Hasil Konversi: {jam} Jam, {menit} Menit, {detik} Detik")
```

```python
# Menghitung nilai akhir mahasiswa

# variabel input nilai mahasiswa
nilai_tugas = float(input("Masukkan nilai tugas: "))
nilai_kuis = float(input("Masukkan nilai kuis: "))
nilai_uts = float(input("Masukkan nilai UTS: "))
nilai_praktikum = float(input("Masukkan nilai praktikum: "))
nilai_uas = float(input("Masukkan nilai UAS: "))

# Menghitung nilai akhir mahasiswa dari 5 komponen nilai
nilai_mahasiswa = (nilai_tugas + nilai_kuis + nilai_uts + nilai_praktikum + nilai_uas) / 5

# Standar kelulusan dengan input keyboard
standar_kelulusan = float(input("Masukkan standar kelulusan: "))

# Proses pengecekan
apakah_lulus = nilai_mahasiswa >= standar_kelulusan
perlu_remedial = nilai_mahasiswa < standar_kelulusan

# Menampilkan hasil output
print(f"Nilai Mahasiswa: {nilai_mahasiswa}")
print(f"Standar Kelulusan: {standar_kelulusan}")
print(f"Apakah Mahasiswa Lulus? {apakah_lulus}")
print(f"Perlu Remedial? {perlu_remedial}")
```

```python
# Program Menghitung Diskon Belanja
belanja = float(input("Masukkan total belanja: "))

# Diskon 35% jika belanja di atas 1.000.000
diskon = (0, 0.35 * belanja)[belanja > 1000000]

total_harga = belanja - diskon
print("Total harga setelah diskon =", total_harga)
```
