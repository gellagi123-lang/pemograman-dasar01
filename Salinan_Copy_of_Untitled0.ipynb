```python
# === LATIHAN 4 ===
tegangan_harian = [217.5, 219.0, 220.3, 218.7, 221.1, 216.9]

# (a) Cetak pembacaan pertama & terakhir pakai indeks
print("Pembacaan pertama:", tegangan_harian[0])
print("Pembacaan terakhir:", tegangan_harian[-1])

# (b) Tambahkan 222.0 pakai append()
tegangan_harian.append(222.0)
print("Setelah ditambah:", tegangan_harian)

# (c) Cetak jumlah total data pakai len()
print("Jumlah total data:", len(tegangan_harian))
```

```python
# === LATIHAN 5 ===
# Buat dictionary spesifikasi motor
motor = {
    "nama": "Motor Induksi 3 Fasa",
    "daya_kW": 5.5,
    "tegangan_V": 380,
    "rpm": 1450
}

# Cetak dictionary awal
print("=== Spesifikasi Motor ===")
print(motor)

# Ubah nilai rpm pakai kunci
motor["rpm"] = 1500

# Cetak ulang setelah diubah
print("\nSetelah nilai rpm diubah:")
print(motor)
```

```python
# === LATIHAN 6 ===
# Data 3 unit pembangkit
pembangkit = [
    {"nama": "Unit-1", "kapasitas_MW": 100, "status": "operasi"},
    {"nama": "Unit-2", "kapasitas_MW": 150, "status": "operasi"},
    {"nama": "Unit-3", "kapasitas_MW": 80, "status": "pemeliharaan"}
]

# (a) Cetak nama unit kedua
print("Nama unit kedua:", pembangkit[1]["nama"])

# (b) Ubah status salah satu unit
pembangkit[0]["status"] = "siaga"
print("\nSetelah ubah status Unit-1:", pembangkit[0])

# (c) Tambahkan unit baru pakai append()
unit_baru = {"nama": "Unit-4", "kapasitas_MW": 200, "status": "operasi"}
pembangkit.append(unit_baru)

# (d) Cetak SEMUA data
print("\n=== Semua Data Pembangkit ===")
for unit in pembangkit:
    print(unit)
```
