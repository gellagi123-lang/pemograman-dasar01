```python
#LIMA LATIHAN KECIL
#1. Data dan skor pertama serta terakhir

scores = [80, 65, 90, 75, 70, 85]

print("Skor pertama:", scores[0])
print("Skor terakhir:", scores[-1])
```

```python
#2. Rata-rata dan jumlah skor di atas rata-rata

average = sum(scores) / len(scores)
print("Rata-rata:", average)

count_above = 0
for score in scores:
    if score > average:
        count_above = count_above + 1
print("Jumlah skor di atas rata-rata:", count_above)
```

```python
#3. Posisi skor terendah

pos_min = 0
for i in range(len(scores)):
    if scores[i] < scores[pos_min]:
        pos_min = i

print("Skor terendah:", scores[pos_min])
print("Posisi skor terendah:", pos_min + 1)
```

```python
#4. List baru: setiap skor ditambah 5

new_scores = []
for score in scores:
    new_scores.append(score + 5)

print(new_scores)
print(scores)
```
